```python
import os
import cv2
import numpy as np
from tqdm import tqdm
```

```python
# =====================================================
# DIRECTORIOS
# =====================================================


#INPUT_DIR_Training  = "../../dataset/Degradacion/ORL_escala4/Training"
#OUTPUT_DIR_Training = "../../dataset/Tecnica De mejora de imagen/ORL_CLAHE_x4/Training"

INPUT_DIR_Testing  = "../../dataset/Degradacion/ORL_escala4/Testing"
OUTPUT_DIR_Testing = "../../dataset/Tecnica De mejora de imagen/ORL_CLAHE_x4/Testing"


#os.makedirs(OUTPUT_DIR_Training, exist_ok=True)
os.makedirs(OUTPUT_DIR_Testing, exist_ok=True)
```

```python
# =====================================================
# CLAHE x4, consideración ante escala x2, no se aplica una máscara para eliminar el fondo, para escala x4 se trata de eliminar un poco el ruido, ya que, estas imágenes
# son las menos reconocidas y de esta manera los modelos de Reconocimiento facial tendrá el desafio de reconocer a los sujetos con estas características
# =====================================================
def pipeline_CLAHE_x4(imagen_lr,
                      clip_limit=2.0,
                      tile_grid_size=(8,8)):

    # Paso 1: Eliminación de ruido
    img_denoised = cv2.fastNlMeansDenoising(
        imagen_lr,
        h=10,
        templateWindowSize=7,
        searchWindowSize=21
    )

    # Paso 2: CLAHE
    clahe = cv2.createCLAHE(
        clipLimit=clip_limit,
        tileGridSize=tile_grid_size
    )

    img_clahe = clahe.apply(img_denoised)

    return img_denoised, img_clahe
```

```python
# =====================================================
# BUSCAR IMÁGENES
# =====================================================

imagenes = []

for root, dirs, files in os.walk(INPUT_DIR_Testing):

    for file in files:

        if file.lower().endswith(".png"):

            imagenes.append(
                os.path.join(root, file)
            )

print(f"Imágenes encontradas: {len(imagenes)}")
```

# Generación del dataset escala x4 CLAHE training

```python
# =====================================================
# GENERAR DATASET CLAHE en escala x4
# =====================================================

for ruta_img in tqdm(imagenes):

    img = cv2.imread(
        ruta_img,
        cv2.IMREAD_GRAYSCALE
    )

    if img is None:
        continue

    _, img_clahe = pipeline_CLAHE_x4(img)

    # Mantener estructura s1,s2,...
    relative_path = os.path.relpath(
        ruta_img,
        INPUT_DIR_Training
    )

    save_path = os.path.join(
        OUTPUT_DIR_Training,
        relative_path
    )

    os.makedirs(
        os.path.dirname(save_path),
        exist_ok=True
    )

    cv2.imwrite(
        save_path,
        img_clahe
    )

print("\nDataset CLAHE escala x4 Training generado correctamente.")
print(f"Total imágenes: {len(imagenes)}")
print(f"Guardado en: {OUTPUT_DIR_Training}")
```

# Generación del dataset escala x4 CLAHE testing

```python
# =====================================================
# GENERAR DATASET CLAHE en escala x4 Testing
# =====================================================

for ruta_img in tqdm(imagenes):

    img = cv2.imread(
        ruta_img,
        cv2.IMREAD_GRAYSCALE
    )

    if img is None:
        continue

    _, img_clahe = pipeline_CLAHE_x4(img)

    # Mantener estructura s1,s2,...
    relative_path = os.path.relpath(
        ruta_img,
        INPUT_DIR_Testing
    )

    save_path = os.path.join(
        OUTPUT_DIR_Testing,
        relative_path
    )

    os.makedirs(
        os.path.dirname(save_path),
        exist_ok=True
    )

    cv2.imwrite(
        save_path,
        img_clahe
    )

print("\nDataset CLAHE escala x4 Testing generado correctamente.")
print(f"Total imágenes: {len(imagenes)}")
print(f"Guardado en: {OUTPUT_DIR_Testing}")
```
